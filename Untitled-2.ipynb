```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
from datetime import datetime
import uuid
```

```python

# Set up plotting style
# plt.style.use('seaborn')
sns.set_palette("husl")
# dburl = f"sqlite:///torqfiskur1.db"
# engine = create_engine(dburl, echo=False, connect_args={"check_same_thread": False})

# Connect to SQLite database
conn = sqlite3.connect('torqfiskur1.db')  # Adjust path to your database

# Load data into pandas DataFrame
query = "SELECT * FROM torqlogs"
df = pd.read_sql_query(query, conn)

# Close connection
conn.close()
```

```python

# Convert Device_Time to datetime
df['Device_Time'] = pd.to_datetime(df['Device_Time'], errors='coerce')

# Basic data cleaning
# Remove duplicates
# df = df.drop_duplicates()

# Handle missing values for key numeric columns
numeric_cols = [
    'GPS_Longitude', 'GPS_Latitude', 'Engine_RPMrpm', 'Speed_OBDkmh',
    'Miles_Per_GallonInstantmpg', 'Trip_average_MPGmpg',
    'Engine_Coolant_TemperatureC', 'Intake_Air_TemperatureC',
    'Horsepower_At_the_wheelshp', ]
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Basic statistics
print("Dataset Overview:")
print(df[numeric_cols].describe())
```

```python

# Create analysis directory
import os
if not os.path.exists('torq_analysis'):
    os.makedirs('torq_analysis')

# Visualization 1: MPG Distribution
plt.figure(figsize=(10, 6))
sns.histplot(df['Miles_Per_GallonInstantmpg'].dropna(), bins=50, kde=True)
plt.title('Distribution of Instant MPG')
plt.xlabel('Miles Per Gallon (Instant)')
plt.ylabel('Frequency')
plt.savefig('torq_analysis/mpg_distribution.png')
plt.close()

# Visualization 2: Speed vs MPG Scatter
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Speed_OBDkmh', y='Miles_Per_GallonInstantmpg', alpha=0.5)
plt.title('Speed vs Instant MPG')
plt.xlabel('Speed (km/h)')
plt.ylabel('Miles Per Gallon (Instant)')
plt.savefig('torq_analysis/speed_vs_mpg.png')
plt.close()

# Visualization 3: Engine RPM vs Horsepower
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Engine_RPMrpm', y='Horsepower_At_the_wheelshp', alpha=0.5)
plt.title('Engine RPM vs Horsepower')
plt.xlabel('Engine RPM')
plt.ylabel('Horsepower (at wheels)')
plt.savefig('torq_analysis/rpm_vs_hp.png')
plt.close()

# Visualization 4: Time series of Speed
plt.figure(figsize=(12, 6))
df_subset = df.dropna(subset=['Device_Time', 'Speed_OBDkmh']).sort_values('Device_Time')
plt.plot(df_subset['Device_Time'], df_subset['Speed_OBDkmh'])
plt.title('Vehicle Speed Over Time')
plt.xlabel('Time')
plt.ylabel('Speed (km/h)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('torq_analysis/speed_over_time.png')
plt.close()

# Visualization 5: Correlation Heatmap
plt.figure(figsize=(12, 8))
correlation_matrix = df[numeric_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Key Metrics')
plt.tight_layout()
plt.savefig('torq_analysis/correlation_heatmap.png')
plt.close()

# Save basic statistics to CSV
stats = df[numeric_cols].describe()
stats.to_csv('torq_analysis/basic_statistics.csv')

# Print top 5 correlations
print("\nTop 5 Correlations:")
corr_unstack = correlation_matrix.unstack()
corr_sorted = corr_unstack.sort_values(kind="quicksort", ascending=False)
# Filter out self-correlations
corr_filtered = corr_sorted[(corr_sorted != 1.0) & (corr_sorted.index.get_level_values(0) != corr_sorted.index.get_level_values(1))]
print(corr_filtered.head(5))

# Calculate some key metrics
avg_mpg = df['Miles_Per_GallonInstantmpg'].mean()
max_speed = df['Speed_OBDkmh'].max()
avg_hp = df['Horsepower_At_the_wheelshp'].mean()

print(f"\nKey Metrics:")
print(f"Average Instant MPG: {avg_mpg:.2f} mpg")
print(f"Maximum Speed: {max_speed:.2f} km/h")
print(f"Average Horsepower: {avg_hp:.2f} hp")

# Save summary report
with open('torq_analysis/summary_report.txt', 'w') as f:
    f.write("TorqLogs Analysis Summary\n")
    f.write("="*30 + "\n")
    f.write(f"Total Records: {len(df)}\n")
    f.write(f"Average Instant MPG: {avg_mpg:.2f} mpg\n")
    f.write(f"Maximum Speed: {max_speed:.2f} km/h\n")
    f.write(f"Average Horsepower: {avg_hp:.2f} hp\n")
    f.write("\nTop 5 Correlations:\n")
    f.write(str(corr_filtered.head(5)))

print("\nAnalysis complete. Visualizations and reports saved in 'torq_analysis' directory.")
```
